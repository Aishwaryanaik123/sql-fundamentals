```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.DataFrame({
    "Study": ["Yes", "Yes", "No", "No"],
    "Pass": ["Yes", "No", "Yes", "No"],
    "Count": [30, 10, 20, 40]
})

data
```

```python
total = data["Count"].sum()

P_A = data[data["Study"] == "Yes"]["Count"].sum() / total

P_A_and_B = data[
    (data["Study"] == "Yes") & (data["Pass"] == "Yes")
]["Count"].sum() / total

P_B = data[data["Pass"] == "Yes"]["Count"].sum() / total

P_A_given_B = P_A_and_B / P_B

print("P(A) =", P_A)
print("P(A ∩ B) =", P_A_and_B)
print("P(A|B) =", P_A_given_B)
```

```python
mean = 100
std = 15

x = np.linspace(mean - 4 * std, mean + 4 * std, 1000)

y = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(
    -0.5 * ((x - mean) / std) ** 2
)

plt.figure(figsize=(10, 5))
plt.plot(x, y)
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.grid()
plt.show()
```

```python
print("68% of values lie within ±1 standard deviation")
print("95% of values lie within ±2 standard deviations")
print("99.7% of values lie within ±3 standard deviations")

print("\nFor mean = 100 and standard deviation = 15:")
print("±1 SD:", mean - std, "to", mean + std)
print("±2 SD:", mean - 2 * std, "to", mean + 2 * std)
print("±3 SD:", mean - 3 * std, "to", mean + 3 * std)
```

```python
n = 10
p = 0.5
simulations = 1000

binomial_data = np.random.binomial(n, p, simulations)

print("Number of simulations:", simulations)
print("Mean number of successes:", binomial_data.mean())
```

```python
values, counts = np.unique(binomial_data, return_counts=True)

plt.figure(figsize=(10, 5))
plt.bar(values, counts)
plt.title("Binomial Distribution Simulation")
plt.xlabel("Number of Successes")
plt.ylabel("Frequency")
plt.xticks(range(n + 1))
plt.show()
```

```python
population = np.random.exponential(scale=10, size=10000)

sample_means = []

for i in range(1000):
    sample = np.random.choice(population, size=30)
    sample_means.append(sample.mean())

sample_means = np.array(sample_means)

print("Population mean:", population.mean())
print("Mean of sample means:", sample_means.mean())
```

```python
plt.figure(figsize=(10, 5))
plt.hist(sample_means, bins=30)
plt.title("Central Limit Theorem - Sampling Distribution")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.show()
```

```python
### Central Limit Theorem

The Central Limit Theorem states that when repeated random samples are taken from a population, the distribution of the sample means approaches a normal distribution as the sample size becomes sufficiently large.

In this simulation, 1000 samples of size 30 were taken from an exponential population. The distribution of the sample means becomes approximately bell-shaped, demonstrating the Central Limit Theorem.
```
