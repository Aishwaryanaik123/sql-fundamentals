```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    "Group": ["A"] * 10 + ["B"] * 10,
    "Score": [
        45, 50, 52, 55, 58,
        60, 62, 65, 68, 70,
        55, 57, 60, 62, 64,
        66, 68, 70, 72, 75
    ]
}

df = pd.DataFrame(data)

df
```

```python
mean = df["Score"].mean()
median = df["Score"].median()
mode = df["Score"].mode().tolist()
variance = df["Score"].var()
standard_deviation = df["Score"].std()

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
print("Variance:", variance)
print("Standard Deviation:", standard_deviation)
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Score"], bins=6, edgecolor="black")

plt.title("Frequency Distribution of Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")

plt.show()
```

```python
print("Distribution Shape: Approximately symmetric")
print("The scores are distributed around the center without a strong left or right skew.")
```

```python
percentile_25 = np.percentile(df["Score"], 25)
percentile_50 = np.percentile(df["Score"], 50)
percentile_75 = np.percentile(df["Score"], 75)
percentile_90 = np.percentile(df["Score"], 90)

print("25th Percentile:", percentile_25)
print("50th Percentile:", percentile_50)
print("75th Percentile:", percentile_75)
print("90th Percentile:", percentile_90)
```

```python
Q1 = df["Score"].quantile(0.25)
Q2 = df["Score"].quantile(0.50)
Q3 = df["Score"].quantile(0.75)

IQR = Q3 - Q1

print("Q1 (25th Percentile):", Q1)
print("Q2 (50th Percentile / Median):", Q2)
print("Q3 (75th Percentile):", Q3)
print("Interquartile Range (IQR):", IQR)
```

```python
print("Interpretation:")
print("The middle 50% of the scores lie between 56.5 and 68.0.")
print("The interquartile range is 11.5, representing the spread of the middle 50% of the data.")
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(
    [df[df["Group"] == "A"]["Score"],
     df[df["Group"] == "B"]["Score"]],
    tick_labels=["Group A", "Group B"]
)

plt.title("Score Distribution by Group")
plt.xlabel("Group")
plt.ylabel("Score")

plt.show()
```

```python
print("Interpretation:")
print("Group A and Group B are compared using their median, spread, and overall distribution.")
print("Group B has a higher median score than Group A.")
print("The box plot shows the median, interquartile range, and possible outliers for each group.")
```
